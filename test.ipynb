```python
import pandas as pd
from pathlib import Path

BASE = Path('/Users/bliekendaal/Desktop/Code')
df = pd.read_parquet(BASE / 'users_met_scores.parquet')
df = df.drop_duplicates(subset='user_id', keep='first')

df['geslacht'] = pd.to_numeric(df['rec_user_gender'], errors='coerce')
df['heeft_geslacht'] = df['geslacht'].notna()

# Hoeveel kolommen hebben ze ingevuld?
score_cols = [c for c in df.columns if c.startswith('rec_')]
df['n_ingevuld'] = df[score_cols].apply(
    lambda col: pd.to_numeric(col, errors='coerce')
).notna().sum(axis=1)

print(f"Met geslacht:    {df['heeft_geslacht'].sum():,} gebruikers, gemiddeld {df[df['heeft_geslacht']]['n_ingevuld'].mean():.1f} velden ingevuld")
print(f"Zonder geslacht: {(~df['heeft_geslacht']).sum():,} gebruikers, gemiddeld {df[~df['heeft_geslacht']]['n_ingevuld'].mean():.1f} velden ingevuld")

# Hebben ze een postcode?
print(f"\nZonder geslacht met postcode: {df[~df['heeft_geslacht']]['postal_code'].notna().sum()}")
print(f"Zonder geslacht met heartrisk: {pd.to_numeric(df[~df['heeft_geslacht']]['rec_heartrisk_cat'], errors='coerce').notna().sum()}")
print(f"Zonder geslacht met BMI: {pd.to_numeric(df[~df['heeft_geslacht']]['rec_med_bmi'], errors='coerce').notna().sum()}")
```

```python
import pandas as pd
from pathlib import Path

BASE = Path('/Users/bliekendaal/Desktop/Code')

df_users = pd.read_parquet(BASE / 'users_mysql.parquet')
df_scores = pd.read_parquet(BASE / 'users_met_scores.parquet')
df_scores = df_scores.drop_duplicates(subset='user_id', keep='first')
actieve_ids = set(df_scores['user_id'].dropna().astype(int))

df_users['jaar'] = pd.to_datetime(df_users['created_at'], errors='coerce').dt.year
df_users['deleted'] = df_users['deleted_at'].notna()
df_users['actief'] = df_users['id'].isin(actieve_ids)

print("Platformgroei (alle accounts inclusief deleted):")
print(df_users.groupby('jaar').size().reset_index(name='alle').to_string())
print("\nAccountactivatie (alleen niet-deleted):")
print(df_users[~df_users['deleted']].groupby('jaar').size().reset_index(name='niet_deleted').to_string())
print("\nDeleted accounts per jaar:")
print(df_users[df_users['deleted']].groupby('jaar').size().reset_index(name='deleted').to_string())
```

```python
# Welke info hebben inactieve accounts?
df_inactief = df_users[~df_users['actief'] & ~df_users['deleted']].copy()
print(f"\nInactieve accounts (niet deleted): {len(df_inactief)}")
print(f"Aanmaakjaar verdeling:")
print(df_inactief['jaar'].value_counts().sort_index().to_string())

# Hebben ze een email domain?
df_inactief['domain'] = df_inactief['email'].str.split('@').str[-1]
print(f"\nTop 10 email domeinen inactief:")
print(df_inactief['domain'].value_counts().head(10).to_string())

# Hebben ze ooit ingelogd?
if 'last_login_at' in df_inactief.columns:
    print(f"\nOoit ingelogd: {df_inactief['last_login_at'].notna().sum()}")
```

```python
import pandas as pd
from pathlib import Path
from sqlalchemy import create_engine

BASE = Path('/Users/bliekendaal/Desktop/Code')
engine = create_engine("postgresql+psycopg2://postgres:barTu6-nidwet-magboh@localhost/smart_health")

df_long   = pd.read_parquet(BASE / 'factor_score_histories.parquet')
df_comp   = pd.read_parquet(BASE / 'completions.parquet')
df_scores = pd.read_parquet(BASE / 'users_met_scores.parquet')
df_scores = df_scores.drop_duplicates(subset='user_id', keep='first')

print("factor_score_histories kolommen:", df_long.columns.tolist())
print(df_long.head(1).to_string())

print("\ncompletions kolommen:", df_comp.columns.tolist())
print(df_comp.head(3).to_string())

# Hoeveel participant_ids overlappen?
overlap = set(df_long['participant_id']) & set(df_comp['participant_id'])
print(f"\nOverlap participant_id long/comp: {len(overlap)}")

# Is store_id beschikbaar via users_met_scores?
print(f"\nstore_id in scores: {'store_id' in df_scores.columns}")
print(f"Unieke stores: {df_scores['store_id'].nunique()}")
print(df_scores['store_id'].value_counts().head(5))
```
